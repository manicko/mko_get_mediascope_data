# Mediascope Jupyter workflow (local JupyterLab / classic)

This notebook demonstrates a repo-shipped, notebook-safe workflow for
`mko_get_mediascope_data` using **CR-3 Alternative A: internal DI workspace bootstrap**
(current architecture, caveated). It is intended for **local JupyterLab or classic
Jupyter Notebook** with Python 3.12+.

## What this notebook does (contract)
1. Installs the package directly from Git (selectable ref, pinned default).
2. Configures portable workspace folders via notebook cell variables: output/export,
   reports source path, and settings/config workspace.
3. Bootstraps/validates a workspace-local settings tree (non-credential templates).
4. Uses an **internal dependency-injected** `AppPaths`/`PathResolver`/`AppService`
   (not the stable public notebook/singleton API) and runs:
   `await service.run_report(Path(REPORT_PATH), progress=NullProgressReporter())`.
5. Verifies export output (file listing/count/sizes only; no sensitive content).

## Readiness / caveats (CR-1, CR-8)
- **Active event loop:** use native `await` in notebook cells. Do **not** use
  `event-loop nesting packages`. Do **not** use the public sync singleton wrapper in an active loop.
- **Internal modules:** `core.paths`/`core.services`/`core.progress` are **not** a
  stable public notebook API. Public `run_report*`, `initialize_settings`, and CLI
  `init`/`list-reports` remain user-dir/singleton oriented.
- **Global logging side effect:** `AppService(...)` applies global
  `logging.config.dictConfig()` during construction (`prepare_app()`). In a
  long-lived kernel, restart the kernel if you need to reset logging/config state.
- **Complete settings tree:** the workspace must contain valid `app_config.json5`,
  `log_config.json5`, required `defaults/` (at least `tv/crosstab.json5`), and the
  selected report file.
- **Log paths:** `prepare_app()` resolves relative log-handler filenames against the
  resolver base and may create workspace log directories.
- **Credentials / safety (CR-7):** credentials must stay in a **user-managed config
  file outside notebook content/cells**. This notebook never embeds or prints secret
  values. The bundled `connections/mediascope.json` template is a placeholder and is
  **never** silently promoted to usable credentials.
- **Blocked vs real E2E (CR-2/CR-5):** without valid external credentials and
  Mediascope network/API access, the run is **BLOCKED** with an explicit reason and
  **no success claim**. Record sanitized evidence in
  `.ai/evidence/01-jupyter-notebook-e2e.md`.
- **Before commit/sharing:** **clear all notebook outputs**. Outputs may contain
  local paths; do not keep credential contents, tokens, auth secrets, or raw API
  payloads.

```python
# C2 - Install package from Git if needed
# Checks if the package is installed; skips install when present (no version/ref check).
# Run this cell, then restart the kernel if packages/imports changed.

import importlib.util
import subprocess
import sys

REPO_URL = 'https://github.com/manicko/mko_get_mediascope_data.git'
GIT_REF = '0743c3d21ed5ac6291ab0a96dfab26db0664c05f'  # selected ref (pinned commit SHA, tag, or branch)

# Optional non-default override examples (uncomment to use):
# GIT_REF = "master"  # moving default branch
# GIT_REF = "vX.Y.Z"  # use only if a tag exists on the remote

FORCE_REINSTALL = False  # set True to reinstall even if the package is already installed

package_name = "mko_get_mediascope_data"
install_spec = f"git+{REPO_URL}@{GIT_REF}"
print(f"Target: {install_spec}")

installed = importlib.util.find_spec(package_name) is not None
if FORCE_REINSTALL:
    print("FORCE_REINSTALL=True -> installation required.")
    do_install = True
elif installed:
    print(f"Package '{package_name}' is already installed; skipping installation.")
    print("To update or change ref, set FORCE_REINSTALL=True (or reinstall manually), then restart the kernel.")
    do_install = False
else:
    print(f"Package '{package_name}' is not installed; installation required.")
    do_install = True

if do_install:
    if "get_ipython" in globals():
        get_ipython().run_line_magic("pip", f'install --upgrade "{install_spec}"')
    else:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "--upgrade", install_spec])
    print("Install finished. Restart the kernel before continuing if packages/imports changed.")
else:
    print("No install needed.")
```

```python
# C3 - Imports and environment info (sanitized)

import json
import platform
import shutil
import sys
from pathlib import Path

import json5
import platformdirs

import mko_get_mediascope_data
from mko_get_mediascope_data.core.paths import APP_DIR as PACKAGE_APP_DIR, APP_NAME, AppPaths, PathResolver
from mko_get_mediascope_data.core.progress import NullProgressReporter
from mko_get_mediascope_data.core.services import AppService

print("Python:", sys.version.split()[0])
print("Platform:", platform.platform())
print("Package version:", getattr(mko_get_mediascope_data, "__version__", "unknown"))
print("Package location (sanitized):", Path(mko_get_mediascope_data.__file__).resolve().parent)
print("Bundled app dir:", PACKAGE_APP_DIR)
```

```python
# C4 - Configurable folder/credential variables (CR-3, CR-7)
# Set variables in this cell. Paths are portable by default; Windows examples are comments only.

# Portable workspace root (override as needed).
WORKSPACE_ROOT = Path.cwd() / ".mediascope_notebook_workspace"
SETTINGS_WORKSPACE = WORKSPACE_ROOT / "settings"   # settings/config workspace
REPORTS_DIR = SETTINGS_WORKSPACE / "reports"       # reports source path (workspace-local)
EXPORT_DIR = WORKSPACE_ROOT / "output"             # output/export

# OQ-2 default example report (workspace-local after bootstrap). Override allowed.
REPORT_FILENAME = "nat_tv_brands_theiss.json5"
REPORT_PATH = REPORTS_DIR / REPORT_FILENAME
# Example override to user config/external path (Windows example, comment only):
# REPORT_PATH = Path(r"%LOCALAPPDATA%\mko_get_mediascope_data\mko_get_mediascope_data\settings\reports\nat_tv_brands_theiss.json5")

# Credentials: external/user-managed file OUTSIDE notebook content. Never embed secrets.
# Portable user-config default:
CREDENTIALS_PATH = Path(platformdirs.user_config_dir(APP_NAME)) / "settings" / "connections" / "mediascope.json"
# Windows example (comment only):
# import os
# CREDENTIALS_PATH = Path(os.environ["LOCALAPPDATA"]) / "mko_get_mediascope_data" / "mko_get_mediascope_data" / "settings" / "connections" / "mediascope.json"

OVERWRITE_WORKSPACE_TEMPLATES = False   # non-destructive default
COPY_CREDENTIALS_INTO_WORKSPACE = False  # explicit opt-in; never overwrites existing workspace creds
DRY_RUN_BOOTSTRAP_ONLY = False           # setup check only - NOT CR-5 success

BUNDLED_SETTINGS_DIR = PACKAGE_APP_DIR / "settings"


print("WORKSPACE_ROOT:", WORKSPACE_ROOT)
print("SETTINGS_WORKSPACE:", SETTINGS_WORKSPACE)
print("REPORTS_DIR:", REPORTS_DIR)
print("EXPORT_DIR:", EXPORT_DIR)
print("REPORT_PATH:", REPORT_PATH)
print("CREDENTIALS_PATH: <external, user-managed; path configured>")
print("COPY_CREDENTIALS_INTO_WORKSPACE:", COPY_CREDENTIALS_INTO_WORKSPACE)
print("DRY_RUN_BOOTSTRAP_ONLY:", DRY_RUN_BOOTSTRAP_ONLY)
```

```python
# C5 - Preflight (no secret values printed)
# Distinguishes real E2E eligibility vs B5 blocked with explicit reason.

KNOWN_PLACEHOLDER_MARKERS = [
    "you username",
    "you password",
    "your_username",
    "your_password",
    "passw",
    "client_id",
    "00000000-0000-0000-0000-000000000000",
]


def credential_status(path: Path) -> dict:
    # Return non-sensitive status only. Never return or print secret values.
    info = {
        "exists": path.exists(),
        "readable": False,
        "placeholder_detected": None,
        "shape_ok": None,
    }
    if not path.exists() or not path.is_file():
        return info
    try:
        text = path.read_text(encoding="utf-8")
        info["readable"] = True
        lower = text.lower()
        info["placeholder_detected"] = any(marker in lower for marker in KNOWN_PLACEHOLDER_MARKERS)
        try:
            data = json5.loads(text)
            info["shape_ok"] = isinstance(data, dict)
        except Exception:
            info["shape_ok"] = False
    except Exception:
        info["readable"] = False
    return info


bundled_ok = BUNDLED_SETTINGS_DIR.exists() and (BUNDLED_SETTINGS_DIR / "app_config.json5").exists()
bundled_report = BUNDLED_SETTINGS_DIR / "reports" / REPORT_FILENAME
report_resolvable = REPORT_PATH.exists() or bundled_report.exists()
cred = credential_status(CREDENTIALS_PATH)

blocked_reason = None
if not bundled_ok:
    blocked_reason = "Bundled package settings templates not found; check installation."
elif DRY_RUN_BOOTSTRAP_ONLY:
    blocked_reason = "DRY_RUN_BOOTSTRAP_ONLY=True (bootstrap/setup check only; not CR-5 success)."
elif not cred["exists"]:
    blocked_reason = (
        "B5 BLOCKED: external credentials file not found at configured CREDENTIALS_PATH "
        "(user-managed; not created by notebook)."
    )
elif cred["placeholder_detected"] is True:
    blocked_reason = (
        "B5 BLOCKED: external credentials appear to contain placeholder values; "
        "provide valid Mediascope credentials outside notebook content."
    )
elif cred["readable"] is False or cred["shape_ok"] is False:
    blocked_reason = (
        "B5 BLOCKED: external credentials file is unreadable or has invalid shape; "
        "fix the user-managed credentials file."
    )
elif not report_resolvable:
    blocked_reason = (
        f"B5 BLOCKED: report not resolvable (workspace target or bundled): {REPORT_FILENAME}"
    )

print("Bundled settings present:", bundled_ok)
print("Report resolvable:", report_resolvable)
print(
    "Credential file exists/readable/placeholder_detected/shape_ok:",
    cred["exists"],
    cred["readable"],
    cred["placeholder_detected"],
    cred["shape_ok"],
)
if blocked_reason:
    print("E2E status: BLOCKED")
    print("Reason:", blocked_reason)
else:
    print("E2E status: ELIGIBLE for authorized local run (real network/API still required).")
```

```python
# C6 - Workspace bootstrap (must run before C7/AppService construction)
# CR-7: excludes credential files from blanket template copy; never overwrites existing
# workspace creds; never silently promotes bundled placeholder.
# CR-3: sets workspace app_config export_folder to absolute EXPORT_DIR preserving other keys.

CREDENTIAL_REL = Path("connections") / "mediascope.json"
WORKSPACE_CREDENTIAL = SETTINGS_WORKSPACE / CREDENTIAL_REL


def is_credential_template(rel: Path) -> bool:
    rel_posix = rel.as_posix().lower()
    name = rel.name.lower()
    return (
        rel_posix.startswith("connections/")
        or "credential" in rel_posix
        or name == "mediascope.json"
        or name == "mediascope_rename_.json"
    )


def copy_non_credential_templates(src_root: Path, dst_root: Path, overwrite: bool = False) -> None:
    dst_root.mkdir(parents=True, exist_ok=True)
    for item in src_root.rglob("*"):
        rel = item.relative_to(src_root)
        target = dst_root / rel
        if item.is_dir():
            target.mkdir(parents=True, exist_ok=True)
            continue
        if is_credential_template(rel):
            continue
        if target.exists() and not overwrite:
            continue
        target.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy2(item, target)


def ensure_workspace_credentials() -> dict:
    # Safe credential handling. Returns non-sensitive status only.
    status = {
        "workspace_has_credentials": WORKSPACE_CREDENTIAL.exists(),
        "copied": False,
        "action": "none",
    }
    WORKSPACE_CREDENTIAL.parent.mkdir(parents=True, exist_ok=True)
    if WORKSPACE_CREDENTIAL.exists():
        status["action"] = "kept_existing_workspace_credentials"
        return status
    if COPY_CREDENTIALS_INTO_WORKSPACE and CREDENTIALS_PATH.exists() and CREDENTIALS_PATH.is_file():
        src_status = credential_status(CREDENTIALS_PATH)
        if src_status["placeholder_detected"] is True:
            status["action"] = "refused_placeholder_external_credentials"
            return status
        shutil.copy2(CREDENTIALS_PATH, WORKSPACE_CREDENTIAL)
        status.update(
            {
                "workspace_has_credentials": True,
                "copied": True,
                "action": "copied_external_credentials_explicit_opt_in",
            }
        )
        return status
    status["action"] = "no_workspace_credentials_created"
    return status


def update_export_folder(app_config_path: Path, export_dir: Path) -> None:
    data = json5.loads(app_config_path.read_text(encoding="utf-8"))
    if not isinstance(data, dict):
        raise TypeError(f"Expected dict in {app_config_path}")
    data["export_folder"] = str(export_dir.resolve())
    app_config_path.write_text(json.dumps(data, ensure_ascii=False, indent=2) + "\n", encoding="utf-8")


# 1) Copy non-credential templates (workspace-local settings tree)
copy_non_credential_templates(
    BUNDLED_SETTINGS_DIR, SETTINGS_WORKSPACE, overwrite=OVERWRITE_WORKSPACE_TEMPLATES
)

# 2) Ensure selected report exists in workspace reports dir (copy bundled default if missing)
REPORTS_DIR.mkdir(parents=True, exist_ok=True)
workspace_report = REPORTS_DIR / REPORT_FILENAME
if not workspace_report.exists() and bundled_report.exists():
    shutil.copy2(bundled_report, workspace_report)

# 3) Safe credential handling (no bundled placeholder promotion)
cred_ws = ensure_workspace_credentials()

# 4) CR-3 export mapping: absolute EXPORT_DIR before AppService construction
EXPORT_DIR.mkdir(parents=True, exist_ok=True)
update_export_folder(SETTINGS_WORKSPACE / "app_config.json5", EXPORT_DIR)

# 5) Minimal completeness checks
required = [
    SETTINGS_WORKSPACE / "app_config.json5",
    SETTINGS_WORKSPACE / "log_config.json5",
    SETTINGS_WORKSPACE / "defaults" / "tv" / "crosstab.json5",
    REPORT_PATH,
]
missing = [str(p) for p in required if not p.exists()]
print("Credential workspace action:", cred_ws["action"])
print("Workspace credentials present:", WORKSPACE_CREDENTIAL.exists())
print("Export folder mapped to absolute:", EXPORT_DIR.resolve())
print("Missing required paths:", missing if missing else "none")
if missing:
    print("Bootstrap incomplete: fix missing paths before constructing AppService.")
```

```python
# C7 - Construct DI AppService (Alt A). Run C6 first.
# Side effect: AppService.prepare_app() applies global logging.config.dictConfig().

app_paths = AppPaths(user_dir=WORKSPACE_ROOT, app_dir=PACKAGE_APP_DIR, app_name=APP_NAME)
resolver = PathResolver(WORKSPACE_ROOT)
service = AppService(app_paths, resolver)

print("Service constructed with workspace-local DI.")
print("Resolved export folder (from service):", service.export_folder)
print("Note: public singleton wrappers and CLI init/list-reports are not used for this Alt A path.")
```

```python
# C8 - Run report (authorized real E2E) or report blocked state
# Requires native `await` (Jupyter/IPython active loop). No event-loop nesting packages.

if blocked_reason:
    print("RUN STATUS: BLOCKED - no success claim.")
    print("Reason:", blocked_reason)
    export_paths = []
else:
    try:
        export_paths = await service.run_report(Path(REPORT_PATH), progress=NullProgressReporter())
        # run_report returns export target paths; individual entries may be skipped (warned).
        print("RUN STATUS: COMPLETED (check warnings/output verification; entries may be skipped).")
        print("Returned export target paths:", [str(p) for p in export_paths])
    except Exception as exc:
        export_paths = []
        # Do not print secrets/auth payloads; show exception type and non-sensitive message.
        print("RUN STATUS: FAILED - no success claim.")
        print(f"{type(exc).__name__}: {exc}")
```

```python
# C9 - Verify/list export files (no sensitive content)
# Do not claim success from path existence alone; run_report may skip entries.

def list_files(root: Path):
    if not root.exists():
        return []
    return sorted([p for p in root.rglob("*") if p.is_file()])


files = list_files(EXPORT_DIR)
print("EXPORT_DIR:", EXPORT_DIR)
print("File count:", len(files))
for p in files[:200]:
    try:
        size = p.stat().st_size
    except OSError:
        size = "?"
    print(f"- {p.relative_to(EXPORT_DIR)} ({size} bytes)")
if len(files) > 200:
    print(f"... ({len(files) - 200} more files omitted)")
if blocked_reason:
    print("Verification note: run was BLOCKED; missing outputs are expected.")
```
